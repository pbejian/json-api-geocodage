## Étape 2 — Compter à la main

```python
for ville in data['villes']:
    print(ville['nom'], ':' , len(ville['restaurants']), "restaurants.")
```

## Étape 3 — Tout mettre à plat

```python
df = pd.json_normalize(data, record_path=["villes", "restaurants"])
df
```

```python
df = pd.json_normalize(
    data,
    record_path=["villes", "restaurants"],
    meta=[["villes", "nom"], ["villes", "region"], ["villes", "habitants"]]
)
df
```

```python
df = df.rename(columns={"villes.nom": "ville",
                        "villes.region": "region",
                        "villes.habitants": "habitants"
                        })
df
```

## Étape 4 — Exploiter

```python
def moyenne(L):
    return round(sum(L)/len(L), 2)
```

```python
df['note_moyenne'] = df['notes'].apply(moyenne)
```

```python
df
```

```python
df = df[['nom', 'type', 'ville', 'region', 'habitants', 'note_moyenne' ]]
df
```

```python
moyennes = df.groupby('ville')['note_moyenne'].mean().sort_values(ascending=False)
print(moyennes)
print("--"*10)
print(moyennes.idxmax())
print("--"*10)
print(moyennes.max())
```
