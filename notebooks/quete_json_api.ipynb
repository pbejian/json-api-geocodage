```python
import requests

def interroger(url, params, headers=None):
    try:
        r = requests.get(url, params=params, headers=headers, timeout=10)
        r.raise_for_status()
        return r.json()
    except requests.exceptions.Timeout:
        print("Trop lent, on abandonne cette requête")
    except requests.exceptions.HTTPError as e:
        print("Le serveur a répondu une erreur :", e)
    except requests.exceptions.RequestException as e:
        print("Problème réseau :", e)
    return None
```

## Étape 2 — Les cinq

```python
adresses = [
    "20 avenue de Ségur 75007 Paris",
    "Grand Place, Lille",
    "1 rue de la Paix, Paris",
    "place Bellecour, Lyon",
    "zzzzz qwerty 99999",
]
```

```python
list_of_dicts = []
for adresse in adresses:
    params={"q": adresse, "limit": 1}
    r = requests.get(url, params=params, headers=headers, timeout=10)
    reponse = r.json() 
    #print("--"*20)
    #print(adresse)
    liste = reponse['features']
    if liste == []:
        dict_adresse = {'adresse': adresse,
                        'label': None,
                        'score': None,
                        'postcode': None,
                        'city' : None
        }
        list_of_dicts.append(dict_adresse)
    else:
        p = reponse['features'][0]['properties']
        label = p['label']
        score = p['score']
        postcode = p['postcode']
        city = p['city']
        dict_adresse = {'adresse': adresse,
                        'label': label,
                        'score': score,
                        'postcode': postcode,
                        'city' : city

        }
        list_of_dicts.append(dict_adresse)
```

# 4 - Géocoder : d'une adresse à un point

```python
import requests

url = "https://api-adresse.data.gouv.fr/search/"
r = requests.get(url, params={"q": "20 avenue de Ségur 75007 Paris", "limit": 1})
reponse = r.json()

import json
print(json.dumps(reponse, ensure_ascii=False, indent=2)[:1000])
```

### ATTENTION

Le piège, et il fait perdre des heures. Dans coordinates, la longitude vient en premier, la latitude ensuite. C'est la norme GeoJSON, et c'est l'inverse de ce que tout le monde dit à l'oral — on dit toujours « latitude, longitude ».

Si tes points se retrouvent au milieu de l'océan ou dans le désert, c'est presque toujours ça. Une seule vérification : en France, la latitude tourne autour de 45 et la longitude autour de 2.

```python
import pandas as pd

df = pd.json_normalize(reponse, record_path="features")
print(df.columns.tolist())
```

```python
df
```

```python
df["lon"] = df["geometry.coordinates"].str[0]
df["lat"] = df["geometry.coordinates"].str[1]
print(df[["properties.label", "lat", "lon"]])
```

### Attention

.str[0] sur une colonne qui contient des listes. L'accesseur .str de pandas ne sert pas qu'aux chaînes : sur une colonne de listes, il prend l'élément à la position demandée. Pratique, et peu connu.

```python
def geocoder_fr(adresse):
    """Renvoie (latitude, longitude) pour une adresse française, ou (None, None)."""
    url = "https://api-adresse.data.gouv.fr/search/"
    try:
        r = requests.get(url, params={"q": adresse, "limit": 1}, timeout=10)
        r.raise_for_status()
        features = r.json()["features"]
        if not features:
            return None, None
        lon, lat = features[0]["geometry"]["coordinates"]
        return lat, lon
    except requests.exceptions.RequestException:
        return None, None
```

```python
print(geocoder_fr("20 avenue de Ségur 75007 Paris"))
print(geocoder_fr("zzzzz qwerty 99999"))
```

### Attention

- Latitude et longitude sont en 'str'. Ne pas oublier de caster en float avant utilisation.
- il faut absolument 1 seconde entre deux requêtes (siinon blocage de l'IP)

```python
import time

def geocoder_monde(adresse):
    """Renvoie (latitude, longitude) n'importe où dans le monde."""
    url = "https://nominatim.openstreetmap.org/search"
    headers = {"User-Agent": "bootcamp-simplon-da/1.0 (prenom.nom@exemple.fr)"}
    params = {"q": adresse, "format": "json", "limit": 1}
    try:
        r = requests.get(url, params=params, headers=headers, timeout=10)
        r.raise_for_status()
        resultats = r.json()
        if not resultats:
            return None, None
        return float(resultats[0]["lat"]), float(resultats[0]["lon"])
    except requests.exceptions.RequestException:
        return None, None
    finally:
        time.sleep(1)
```

```python
import time
debut = time.time()
print(geocoder_monde("54 Via Pietro Mascagni, Catania, Italy"))
print(geocoder_monde("Piazza del Duomo, Catania, Italy"))
fin = time.time()
print(f"durée totale : {(fin - debut):.1f} s")
```
