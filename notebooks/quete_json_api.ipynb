```python
import requests

def interroger(url, params, headers=None):
    try:
        r = requests.get(url, params=params, headers=headers, timeout=10)
        r.raise_for_status()
        return r.json()
    except requests.exceptions.Timeout:
        print("Trop lent, on abandonne cette requête")
    except requests.exceptions.HTTPError as e:
        print("Le serveur a répondu une erreur :", e)
    except requests.exceptions.RequestException as e:
        print("Problème réseau :", e)
    return None
```

```python
print(interroger("https://api-adresse.data.gouv.fr/search/", {"q": "Lille", "limit": 1}) is not None)
print(interroger("https://api-adresse.data.gouv.fr/nimportequoi", {"q": "Lille"}))
```

```python
adresses = [
    "20 avenue de Ségur 75007 Paris",
    "Grand Place, Lille",
    "1 rue de la Paix, Paris",
    "place Bellecour, Lyon",
    "zzzzz qwerty 99999",
]
```

## Étape 1 — Une seule, d'abord

```python
from pprint import pprint

url = "https://api-adresse.data.gouv.fr/search/"
headers = {"User-Agent": "bootcamp-simplon-da/1.0 (prenom.nom@exemple.fr)"}
params={"q": "20 avenue de Ségur 75007 Paris", "limit": 1}
r = requests.get(url, params=params, headers=headers)
print(r.status_code)
reponse = r.json()
pprint(reponse['features'][0]['properties'])
print("--"*20)
print(reponse['features'][0]['properties']['label'])
print(reponse['features'][0]['properties']['score'])
```

## Étape 2 — Les cinq

```python
adresses = [
    "20 avenue de Ségur 75007 Paris",
    "Grand Place, Lille",
    "1 rue de la Paix, Paris",
    "place Bellecour, Lyon",
    "zzzzz qwerty 99999",
]
```

```python
list_of_dicts = []
for adresse in adresses:
    params={"q": adresse, "limit": 1}
    r = requests.get(url, params=params, headers=headers, timeout=10)
    reponse = r.json() 
    #print("--"*20)
    #print(adresse)
    liste = reponse['features']
    if liste == []:
        dict_adresse = {'adresse': adresse,
                        'label': None,
                        'score': None,
                        'postcode': None,
                        'city' : None
        }
        list_of_dicts.append(dict_adresse)
    else:
        p = reponse['features'][0]['properties']
        label = p['label']
        score = p['score']
        postcode = p['postcode']
        city = p['city']
        dict_adresse = {'adresse': adresse,
                        'label': label,
                        'score': score,
                        'postcode': postcode,
                        'city' : city

        }
        list_of_dicts.append(dict_adresse)
```

```python
pprint(list_of_dicts)
```

## Étape 3 — En DataFrame

```python
df_adresses = pd.json_normalize(list_of_dicts)
df_adresses
```

## Étape 4 — lire le résultat

```python
nb_adresses_trouvee = df_adresses['label'].count()
print("Nombre d'adresses trouvées :" ,nb_adresses_trouvee)
```

```python
print("Score minimium :", df_adresses['score'].min())
index_min =  df_adresses['score'].idxmin()
print("Index du score min :", index_min)
print("Adresse avec score minimium :", df_adresses.loc[index_min]['adresse'])
```

**Pourquoi ce score bas ?** Requête vague : ni numéro, ni code postal. 

**Le label correspond-il toujours ?** Non. L'API renvoie le résultat le plus proche sans prévenir : il faut toujours comparer l'adresse demandée et le label.
